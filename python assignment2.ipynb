```python
my_set={'a','e','i','o','u','a','a','i'} #set automatically removes duplicate
print(my_set)                            #input has 8 values but 'a' repaeted 3 times and 'i' repeated 2 times
                                        #so output will be only 5 unique values
                                        #order may be different as set is unordered
```

```python
my_set[4]='s' 
print(my_set) 
                  #Set is unordered - there is no index 4, so you can't access by position like list
                  #Set is mutable for adding/removing whole elements, but you cannot change an element by index.
                  #To add 's', you must use my_set.add('s')
```

```python
set1={1,3,5,7,9}
set2={2,3,5,8,10}
print("set1:",set1)
print("set2:",set2)
```

```python
print("union:",set1.union(set2))
print("intersection",set1.intersection(set2))
```
